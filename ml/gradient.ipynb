```python
import numpy as np 
import matplotlib.pyplot as plt
```

```python
def funtion(x) : 
    return (x+3)**2 

def derivative(x) : 
    return 2*(x+3) 
```

```python
x = 2 

learing_rate = 0.1 

iterations = 50 

x_values = [] 
y_values = [] 


for i in range(iterations) : 

    x_values.append(x)
    y_values.append(funtion(x)) 


    gradient = derivative(x) 


    x = x - learing_rate*gradient 


    print("Iteration : " , i+1 ,  "x :" , x , "y :" , funtion(x)) 




print("Local Minima :") 

print("x = " , x)

print("y = " , funtion(x))



```

```python
x_range = np.linspace(-5, 4, 100)
y_range = funtion(x_range)

plt.plot(x_range, y_range, label='y = (x + 3)^2')
plt.scatter(x_values, y_values, color='red', label='Gradient Descent')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Gradient Descent')
plt.legend()
plt.grid()
plt.show()
```
